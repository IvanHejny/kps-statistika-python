Spojenci za 2. světové války odhadovali počet vyrobených tanků ze sériových čísel zajatých tanků. Tady si na počítači vyzkoušíme, jak dobře fungují různé odhady.

## Krok 1: Jeden pokus

V simulaci známe skutečný počet tanků `N` (ve skutečnosti ho neznáme, právě ten chceme odhadnout). Počítač „zajme“ `k` náhodných tanků a my spočítáme odhady. Spusťte buňku několikrát a sledujte, jak se výsledek mění.

```python
import random

N = 250   # "skutečný" počet tanků (v simulaci ho smíme znát)
k = 6     # kolik tanků zajmeme

zajate = random.sample(range(1, N + 1), k)   # k náhodných různých čísel od 1 do N
print("Sériová čísla zajatých tanků:", sorted(zajate))

M = max(zajate)          # základní odhad: nejvyšší viděné číslo
opraveny = M + M / k     # opravený odhad
print("Základní odhad M:", M)
print("Opravený odhad M + M/k:", round(opraveny))
```

**Otázka:** Může být základní odhad `M` někdy větší než skutečné `N`? Proč je `M` většinou menší než `N`?

## Krok 2: Mnoho pokusů

Jeden pokus nic nedokazuje, může nám padnout štěstí i smůla. Pokus proto zopakujeme 10 000krát a spočítáme, kam se odhady v průměru trefují a jak velkou chybu typicky děláme.

**Úkol:** Doplňte svůj vlastní odhad. Na řádcích začínajících `#` s tečkami `...` odstraňte znak `#` a místo `...` doplňte výpočet (vzor najdete na řádku nad nimi). Pak buňku spusťte.

```python
import random

N = 250               # "skutečný" počet tanků — v simulaci to smíme znát,
                       # abychom mohli zpětně zkontrolovat naše odhady
k = 6                  # kolik tanků zajmeme v jednom pokusu
pocet_pokusu = 10000

soucet_chyb_M = 0          # sem budeme postupně přičítat chyby základního odhadu M
soucet_chyb_opraveny = 0   # a sem chyby opraveného odhadu M + M/k
soucet_zakladni = 0        # sem budeme přičítat samotné hodnoty M (ne chyby!)
soucet_opraveny = 0        # a sem samotné hodnoty opraveného odhadu
# soucet_chyb_vlastni = ... # a sem chyby VAŠEHO odhadu — doplňte podobně jako výše

for pokus in range(pocet_pokusu):  # pro každý pokus 1, 2, 3, ... až pocet_pokusu:
    # 1. "zajmeme" k náhodných tanků - vybereme k = 6 náhodných různých čísel z 1 až N
    zajate = random.sample(range(1, N + 1), k)

    # 2a. spočítáme základní odhad M = nejvyšší viděné číslo
    M = max(zajate)

    # 2b. spočítáme opravený odhad
    opraveny = M + M / k

    # 2c. váš vlastní odhad — doplňte svůj vlastní vzorec (kombinaci M a k)
    # vlastni = ...

    # 3. o kolik jsme se v tomhle pokusu spletli?
    chyba_M = abs(M - N)
    chyba_opraveny = abs(opraveny - N)
    # chyba_vlastni = ...

    # 4. přičteme chybu do celkového součtu
    soucet_chyb_M = soucet_chyb_M + chyba_M
    soucet_chyb_opraveny = soucet_chyb_opraveny + chyba_opraveny
    # soucet_chyb_vlastni = ...

    # 5. přičteme i samotné odhady (ne chyby) - abychom viděli, kam se v průměru "trefujeme"
    soucet_zakladni = soucet_zakladni + M
    soucet_opraveny = soucet_opraveny + opraveny

# Po všech pokusech: Kam se naše odhady v průměru trefují? (skutečné N je 250)
prumerny_zakladni_odhad = soucet_zakladni / pocet_pokusu
prumerny_opraveny_odhad = soucet_opraveny / pocet_pokusu
print("Průměr základních odhadů M:", prumerny_zakladni_odhad)
print("Průměr opravených odhadů:", prumerny_opraveny_odhad)

# Jak velkou chybu děláme "typicky"?
print("Průměrná absolutní chyba u základního odhadu M:", soucet_chyb_M / pocet_pokusu)
print("Průměrná absolutní chyba u opraveného odhadu:", soucet_chyb_opraveny / pocet_pokusu)
# print("Průměrná chyba u vlastního odhadu:", ...)
```
